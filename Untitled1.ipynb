# Análisis de Pacientes en Servicios de Emergencia

## Problema
En los servicios de urgencias, es fundamental priorizar la atención de los pacientes según su gravedad. Este proyecto analiza datos simulados para identificar patrones que ayuden a mejorar la toma de decisiones.

## Objetivo
Analizar datos utilizando NumPy y Pandas para identificar factores que influyen en la gravedad de los pacientes.

Importar librerías

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

##Crear base de datos simulados

```python

np.random.seed(42)

n = 100  # número de pacientes

edades = np.random.randint(0, 90, n)
gravedad = np.random.randint(1, 6, n)
tiempo_espera = np.random.randint(5, 120, n)

tipo_emergencia = np.random.choice(
    ["Trauma", "Cardiaco", "Respiratorio", "Accidente"],
    n
)

resultado = np.random.choice(
    ["Atendido", "Hospitalizado", "Fallecido"],
    n,
    p=[0.6, 0.3, 0.1]  # probabilidades
)

secuelas = np.random.choice(
    ["Sí", "No"],
    n,
    p=[0.2, 0.8]
)
```

Crear DataFrame


```python
df = pd.DataFrame({
    "Edad": edades,
    "Gravedad": gravedad,
    "Tiempo_Espera": tiempo_espera,
    "Tipo_Emergencia": tipo_emergencia,
    "Resultado": resultado,
    "Secuelas": secuelas
})

df.head()
```

##Pacientes Fallecidos


```python
fallecidos = df[df["Resultado"] == "Fallecido"]
fallecidos
```

##Cantidad de Resultados


```python
df["Resultado"].value_counts()
```

##Pacientes Con Secuelas

```python
df["Secuelas"].value_counts()
```

##Relación Gravedad-Resultado

```python
pd.crosstab(df["Gravedad"], df["Resultado"])
```

##Promedio de Edad- Resultado

```python
df.groupby("Resultado")["Edad"].mean()
```

#Gráfico


```python
df["Resultado"].value_counts().plot(kind="bar")
plt.title("Resultados de los pacientes")
plt.xlabel("Resultado")
plt.ylabel("Cantidad")
plt.show()
```

#Gráfico Gravedad vs Resultado

```python
pd.crosstab(df["Gravedad"], df["Resultado"]).plot(kind="bar", stacked=True)
plt.title("Gravedad vs Resultado")
plt.xlabel("Gravedad")
plt.ylabel("Cantidad")
plt.show()
```

#Conclusiones

Se identificó que los pacientes con mayor nivel de gravedad tienen mayor probabilidad de hospitalización o fallecimiento. Además, se evidenció que algunos pacientes presentan secuelas posteriores a la atención, lo cual es relevante para el seguimiento clínico. Este análisis demuestra cómo el uso de herramientas como NumPy y Pandas permite comprender mejor los datos y apoyar la toma de decisiones en el área de salud.
